# 01 – Data Preparation

This notebook covers data loading, filtering, cleaning,
feature selection, and sampling for customer segmentation.

The original dataset is proprietary and is not included
in this repository.

```python
# Import Libraries
import pandas as pd
import numpy as np
from pathlib import Path

warnings.filterwarnings('ignore')
```

```python
# Load dataset
DATA_PATH = Path("../data/customers.csv")

df = pd.read_csv(DATA_PATH)

print(f"Dataset shape: {df.shape}")
```

```python
# Select completed sales
df_filtered = df.loc[
    df["order_status"] == "Sale"
].copy()

print(f"Filtered dataset shape: {df_filtered.shape}")
```

```python
# Select relevant features
selected_features = [
    "customer_age",
    "contract_duration",
    "monthly_fee",
    "upfront_payment",
    "sim_only",
    "provider_category",
    "device_category",
    "profitability"
]

df_selected = df_filtered[selected_features].copy()
```

```python
# Remove invalid customer ages
df_clean = df_selected.loc[
    df_selected["customer_age"].between(18, 100)
].copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Check missing values
missing_values = df_clean.isna().sum()
print(missing_values)
```

```python
# Export prepared dataset
output_path = Path("../data/customers_prepared.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)

df_clean.to_csv(output_path, index=False)
```
